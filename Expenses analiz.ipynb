```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("expenses.csv")
```

```python
kategori_toplam=df.groupby("Kategori")["Miktar"].sum().sort_values(ascending=False)
```

```python

plt.figure(figsize=(8,8))
wedges, texts, autotexts = plt.pie(
    kategori_toplam.values,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.8,
    explode=[0.05 if v < kategori_toplam.sum()*0.05 else 0 for v in kategori_toplam.values]
)

# Etiketleri dışarı yaz
plt.legend(
    wedges,
    kategori_toplam.index,
    title="Kategoriler",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1)
)

plt.title("Harcama Dağılımı (%)")
plt.tight_layout()
plt.show()
```
